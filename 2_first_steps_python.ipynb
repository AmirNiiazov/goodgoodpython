```python
import math

x0, y0, x1, y1 = map(float, input().split())

dot = x0*x1 + y0*y1
len0 = math.hypot(x0, y0)
len1 = math.hypot(x1, y1)

cos_a = dot / (len0 * len1)
alpha = round(math.degrees(math.acos(cos_a)), 1)
print(alpha)
```
